### Creating Transaction Type

A new TransactionType column will be created to distinguish normal sales from cancelled or returned transactions.

```python
# Classify each transaction as a cancellation or regular sale
df["TransactionType"] = np.where(
    df["InvoiceNo"].str.startswith("C"),
    "Cancellation/Return",
    "Sale"
)
```

```python
# Check the number of transactions in each category
df["TransactionType"].value_counts()
```

```python
# Preview the new TransactionType column
df[['InvoiceNo', 'Quantity', 'TransactionType']].head(20)
```

### Creating Product Category Column

A new ProductCategory column will be created to distinguish products from non products or gift vouchers.

```python
# Stock codes that do not represent normal retail products
non_product_codes = [
    'AMAZONFEE',     # Amazon marketplace fee
    'B',             # Bad debt adjustment
    'BANK CHARGES',  # Bank charges
    'C2',            # Carriage charge
    'CRUK',          # CRUK commission
    'D',             # Discount
    'DOT',           # Dotcom postage
    'M',             # Manual transaction / adjustment
    'POST',          # Postage
    'S',             # Samples
    'ADJUST',        # Manual adjustment
    'ADJUST2',       # Manual adjustment
    'TEST001',       # Test transaction
    'DCGSSGIRL',     # Non-standard / update entry
    'DCGSSBOY'       # Non-standard / update entry
]

# Create a new column and assume transactions are products by default
df['TransactionCategory'] = 'Product'

# Classify known non-product stock codes
df.loc[
    df['StockCode'].isin(non_product_codes),
    'TransactionCategory'
] = 'Non-Product'

# Gift vouchers are revenue but are not normal catalogue products
df.loc[
    df['StockCode'].str.startswith('gift_0001_', na=False),
    'TransactionCategory'
] = 'Gift Voucher'

# Check the classifications
df['TransactionCategory'].value_counts()
```

```python
# View the unique non-product transactions
df.loc[
    df['TransactionCategory'] != 'Product',
    ['StockCode', 'Description', 'TransactionCategory']
].drop_duplicates().sort_values('StockCode')
```

### Convert InvoiceDate to Datetime

InvoiceDate will be converted to datetime format. This will allow time-based analysis to be performed, including monthly sales trends and customer purchasing behaviour over time.

```python
# Convert InvoiceDate to datetime format
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
```

```python
# Confirm that InvoiceDate has been converted successfully
df['InvoiceDate'].dtype
```

InvoiceDate was successfully converted to datetime format, making the column suitable for time-based analysis.

### Convert Identifier Data Types

CustomerID represents a customer identifier rather than a numerical measurement. Therefore, it will be converted to a string data type. InvoiceNo and StockCode will also be treated as identifiers to ensure that they are not interpreted as continuous numerical variables.

```python
# Convert identifier columns to string data types
df['CustomerID'] = df['CustomerID'].astype(str)
df['InvoiceNo'] = df['InvoiceNo'].astype(str)
df['StockCode'] = df['StockCode'].astype(str)
```

```python
# Check the data types after conversion
df.dtypes
```

The identifier columns are now stored as text rather than being treated as numerical measurements. Quantity and UnitPrice remain numerical because they are required for calculations.

### Create Total Transaction Column

A new TotalTransactionValue column will be created by multiplying Quantity by UnitPrice.

Positive values represent regular sales, while negative values represent cancellations or returns.

```python
# Calculate transaction value and round to two decimal places
df['TotalTransactionValue'] = (
    df['Quantity'] * df['UnitPrice']
).round(2)
```

```python
# Preview the new transaction value column
df[['Quantity', 'UnitPrice', 'TransactionType', 'TotalTransactionValue']].head(20)
```

```python
# Confirm that cancellations produce negative transaction values
df[df['TransactionType'] == 'Cancellation/Return'][
    ['Quantity', 'UnitPrice', 'TransactionType', 'TotalTransactionValue']
].head(20)
```

TotalTransactionValue was calculated successfully. Regular sales produce positive transaction values, while cancellations produce negative values. 

### Create Date Columns

Additional date columns will be extracted from InvoiceDate for time-based analysis. Year, Month and YearMonth will be created so that sales and customer spending can be analysed over different time periods.

```python
# Extract the transaction year from InvoiceDate
df['Year'] = df['InvoiceDate'].dt.year
```

```python
# Extract the transaction month from InvoiceDate
df['Month'] = df['InvoiceDate'].dt.month
```

```python
# Create a YearMonth variable for monthly trend analysis
df['YearMonth'] = df['InvoiceDate'].dt.to_period('M').astype(str)
```

```python
# Preview the newly created date features
df[
    ['InvoiceDate', 'Year', 'Month', 'YearMonth']
].head()
```

Year, Month and YearMonth were successfully created from InvoiceDate. These features will support later analysis of yearly and monthly sales trends, with YearMonth providing a chronological monthly variable across both 2010 and 2011.
